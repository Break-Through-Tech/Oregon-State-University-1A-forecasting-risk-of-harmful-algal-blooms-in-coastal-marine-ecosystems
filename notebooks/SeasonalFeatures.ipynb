```python

import pandas as pd
import numpy as np

train = pd.read_csv("data/processed/hab_train.csv")
val = pd.read_csv("data/processed/hab_validation.csv")

def add_seasonal_features(df):
    df = df.copy()
    dates = pd.to_datetime(df["observation_date"])

    df["month"] = dates.dt.month
    df["day_of_year"] = dates.dt.dayofyear

    df["sin_month"] = np.sin(
        2 * np.pi * (df["month"] - 1) / 12
    )
    df["cos_month"] = np.cos(
        2 * np.pi * (df["month"] - 1) / 12
    )

    return df

train = add_seasonal_features(train)
val = add_seasonal_features(val)

print(train[[
    "observation_date", "region", "month",
    "day_of_year", "sin_month", "cos_month"
]].head())

train.to_csv("data/processed/hab_train_seasonal.csv", index=False)
val.to_csv("data/processed/hab_validation_seasonal.csv", index=False)
```
