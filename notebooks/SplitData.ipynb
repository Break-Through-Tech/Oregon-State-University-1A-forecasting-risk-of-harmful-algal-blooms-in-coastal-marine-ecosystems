```python
import pandas as pd

model_df = pd.read_csv(
    "data/processed/hab_model_dataset_initial.csv"
)

env_df = pd.read_csv(
    "data/processed/hab_events_with_environmental_features.csv"
)

print("Model dataset:", model_df.shape)
print("Environmental dataset:", env_df.shape)
```

```python
print("MODEL DATASET COLUMNS:")
print(model_df.columns.tolist())

print("\nENVIRONMENTAL DATASET COLUMNS:")
print(env_df.columns.tolist())
```

```python
display(model_df.head())
display(env_df.head())
```

```python
print(model_df["target"].value_counts())

print("\nTarget percentages:")
print(model_df["target"].value_counts(normalize=True) * 100)

print("\nDate range:")
print(model_df["observation_date"].min())
print(model_df["observation_date"].max())
```

```python
print("Sample types:")
print(model_df["sample_type"].value_counts(dropna=False))

print("\nSynthetic records:")
print(model_df["is_synthetic"].value_counts(dropna=False))

print("\nMatched positive IDs:")
print(model_df["matched_positive_record_id"].notna().sum())

print("\nMissing dates:")
print(model_df["observation_date"].isna().sum())

print("\nDuplicate record IDs:")
print(model_df["record_id"].duplicated().sum())
```

```python
model_df["observation_date"] = pd.to_datetime(
    model_df["observation_date"],
    errors="coerce"
)

print("Earliest date:", model_df["observation_date"].min())
print("Latest date:", model_df["observation_date"].max())

print("\nObservations by year:")
print(
    model_df["observation_date"]
    .dt.year
    .value_counts()
    .sort_index()
)
```

```python
# Separate real HAB events and synthetic non-events
positive_df = model_df[model_df["target"] == 1].copy()
negative_df = model_df[model_df["target"] == 0].copy()

print("Positive dates by year:")
print(positive_df["observation_date"].dt.year.value_counts().sort_index())

print("\nNegative dates by year:")
print(negative_df["observation_date"].dt.year.value_counts().sort_index())

print("\nExample synthetic records:")
display(
    negative_df[
        ["record_id", "matched_positive_record_id",
         "observation_date", "synthetic_method"]
    ].head(10)
)
```

```python
positive_ids = set(positive_df["record_id"])

matched_ids = set(
    negative_df["matched_positive_record_id"].dropna()
)

print("Synthetic parent IDs found in positive records:",
      len(matched_ids & positive_ids))

print("Unique synthetic parent IDs:", len(matched_ids))

print("Unmatched parent IDs:", len(matched_ids - positive_ids))
```

```python
from sklearn.model_selection import train_test_split

# Assign a common group ID to each matched pair
model_df["group_id"] = model_df["record_id"]

mask = model_df["target"] == 0

model_df.loc[mask, "group_id"] = (
    model_df.loc[mask, "matched_positive_record_id"]
)

# Get unique matched pairs
groups = model_df["group_id"].unique()

# Split groups, not individual rows
train_groups, val_groups = train_test_split(
    groups,
    test_size=0.20,
    random_state=42
)

# Create training and validation datasets
train_df = model_df[
    model_df["group_id"].isin(train_groups)
].copy()

val_df = model_df[
    model_df["group_id"].isin(val_groups)
].copy()

print("Training:", train_df.shape)
print("Validation:", val_df.shape)
```

```python
# Check class balance
print("Training target counts:")
print(train_df["target"].value_counts())

print("\nValidation target counts:")
print(val_df["target"].value_counts())

# Check that no matched pairs appear in both sets
overlap = set(train_df["group_id"]) & set(val_df["group_id"])

print("\nOverlapping groups:", len(overlap))

assert len(overlap) == 0
assert len(train_df) + len(val_df) == len(model_df)
```

```python
train_df = train_df.drop(columns=["group_id"])
val_df = val_df.drop(columns=["group_id"])

train_df.to_csv(
    "data/processed/hab_train.csv",
    index=False
)

val_df.to_csv(
    "data/processed/hab_validation.csv",
    index=False
)

print("Training and validation datasets saved!")
```
