```python
import re
import numpy as np

def parse_temperature(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    # Extract numbers without treating range hyphens as negative signs
    numbers = re.findall(r"\d+(?:\.\d+)?", text)

    if len(numbers) == 0:
        return np.nan

    numbers = [float(x) for x in numbers]

    # Single temperature
    if len(numbers) == 1:
        return numbers[0]

    # Temperature range -> midpoint
    if len(numbers) == 2:
        return sum(numbers) / 2

    return np.nan
```

```python
df["Water Temp Celsius"] = df["Water Temp"].apply(parse_temperature)
```

```python
import re
import numpy as np

def parse_salinity(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    # Match numbers including decimals such as .04
    numbers = re.findall(r"(?:\d+\.\d+|\.\d+|\d+)", text)

    if len(numbers) == 0:
        return np.nan

    numbers = [float(x) for x in numbers]

    # If "at Xm" appears, the first number is the salinity
    if re.search(r"\bat\s*\d*\.?\d+\s*m\b", text):
        return numbers[0]

    # One number = straightforward value
    if len(numbers) == 1:
        return numbers[0]

    # Only treat two numbers as a range if a range separator is present
    if re.search(r"\d\s*(?:-|–|to)\s*\d", text):
        return (numbers[0] + numbers[1]) / 2

    # Too ambiguous to safely interpret
    return np.nan
```

```python
df["Salinity Standardized"] = df["Salinity"].apply(parse_salinity)
```

```python
import re
import numpy as np

def parse_oxygen(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    # Non-numeric descriptions
    if text in ["no", "oxygen depletion", "n.a.", "na"]:
        return np.nan

    # Remove mg/L-style units
    cleaned = re.sub(r"mg\s*/?\s*l(?:-1)?", "", text)
    cleaned = cleaned.strip()

    # Single numeric value
    if re.fullmatch(r"<?\s*(?:\d+\.\d+|\.\d+|\d+)", cleaned):
        numbers = re.findall(r"(?:\d+\.\d+|\.\d+|\d+)", cleaned)
        return float(numbers[0])

    # Simple range such as 1.47 - 10.55 or 6-8
    range_match = re.fullmatch(
        r"\s*(\d*\.?\d+)\s*-\s*(\d*\.?\d+)\s*",
        cleaned
    )

    if range_match:
        low = float(range_match.group(1))
        high = float(range_match.group(2))
        return (low + high) / 2

    # Everything more complicated stays missing
    return np.nan
```

```python
df["Oxygen Content Standardized"] = (
    df["Oxygen Content"].apply(parse_oxygen)
)
```

```python
def parse_oxygen_saturation(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    # Text/non-measurements
    if text in ["no", "n.a.", "na"]:
        return np.nan

    # Don't guess inequality values
    if "<" in text or ">" in text:
        return np.nan

    # Multiple depth measurements are too complex to reduce safely
    if ";" in text:
        return np.nan

    # Remove percent sign
    cleaned = text.replace("%", "").strip()

    # Simple range
    range_match = re.fullmatch(
        r"(\d*\.?\d+)\s*-\s*(\d*\.?\d+)",
        cleaned
    )

    if range_match:
        low = float(range_match.group(1))
        high = float(range_match.group(2))
        return (low + high) / 2

    # Single measurement, optionally followed by depth information
    number_match = re.match(r"^(\d*\.?\d+)", cleaned)

    if number_match:
        return float(number_match.group(1))

    return np.nan
```

```python
df["Oxygen Saturation Standardized"] = (
    df["Oxygen Saturation"].apply(parse_oxygen_saturation)
)
```

```python
def parse_turbidity(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    # Non-numeric response
    if text in ["no", "n.a.", "na"]:
        return np.nan

    # Inequalities don't give an exact measurement
    if "<" in text or ">" in text:
        return np.nan

    # Simple range -> midpoint
    range_match = re.fullmatch(
        r"(\d*\.?\d+)\s*-\s*(\d*\.?\d+)",
        text
    )

    if range_match:
        low = float(range_match.group(1))
        high = float(range_match.group(2))
        return (low + high) / 2

    # Single number
    if re.fullmatch(r"\d*\.?\d+", text):
        return float(text)

    return np.nan
```

```python
df["Turbidity Standardized"] = (
    df["Turbidity"].apply(parse_turbidity)
)
```

```python
def parse_secchi(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    # Non-measurement
    if text in ["no", "n.a.", "na"]:
        return np.nan

    # Explicit centimeters -> convert to meters
    cm_match = re.fullmatch(r"(\d*\.?\d+)\s*cm", text)
    if cm_match:
        return float(cm_match.group(1)) / 100

    # Extract numbers
    numbers = re.findall(r"(?:\d+\.\d+|\.\d+|\d+)", text)
    numbers = [float(x) for x in numbers]

    if len(numbers) == 0:
        return np.nan

    # One measurement
    if len(numbers) == 1:
        return numbers[0]

    # Two measurements representing a range/min-max
    if len(numbers) == 2:
        if (
            "-" in text
            or " to " in text
            or ("min" in text and "max" in text)
        ):
            return sum(numbers) / 2

    return np.nan
```

```python
df["Secchi Disc Meters"] = df["Secchi Disc"].apply(parse_secchi)
```

```python
def parse_current_velocity(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    if text in ["n.a.", "na", "no", "slack"]:
        return np.nan

    try:
        return float(text)
    except ValueError:
        return np.nan
```

```python
def parse_wind_velocity(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    # Qualitative / non-exact measurements
    if text in ["no", "calm", "light", "high", "very low", "n.a.", "na"]:
        return np.nan

    # Beaufort force descriptions
    if "force" in text:
        return np.nan

    # Inequalities are not exact values
    if "<" in text or ">" in text or "less than" in text:
        return np.nan

    numbers = re.findall(r"(?:\d+\.\d+|\.\d+|\d+)", text)
    numbers = [float(x) for x in numbers]

    if len(numbers) == 0:
        return np.nan

    # If range, use midpoint
    if len(numbers) == 2:
        value_num = sum(numbers) / 2
    elif len(numbers) == 1:
        value_num = numbers[0]
    else:
        return np.nan

    # Knots -> m/s
    if "knot" in text or "kts" in text:
        return value_num * 0.514444

    # mph -> m/s
    if "mph" in text:
        return value_num * 0.44704

    # km/h -> m/s
    if "km" in text and ("h" in text or "hour" in text):
        return value_num / 3.6

    # Explicit m/s variants
    if (
        "m/s" in text
        or "m/sec" in text
        or "m/second" in text
        or "m/seg" in text
        or "ms-1" in text
    ):
        return value_num

    # Plain numeric values: preserve as source-reported numeric values
    if re.fullmatch(r"\d*\.?\d+", text):
        return value_num

    # Plain numeric range
    if re.fullmatch(r"\d*\.?\d+\s*-\s*\d*\.?\d+", text):
        return value_num

    return np.nan
```

```python
df["Wind Velocity Standardized"] = (
    df["Wind Velocity"].apply(parse_wind_velocity)
)
```
